```python
# ---------- گرفتن نمرات ----------
scores = []

n = int(input("چند تا دانش‌آموز داری؟ "))

for i in range(n):
    score = int(input("نمره‌ی دانش‌آموز شماره " + str(i + 1) + ": "))
    scores.append(score)

# ---------- محاسبات ----------
total = 0
passed = 0
failed = 0

for score in scores:
    total += score
    if score >= 10:
        passed += 1
    else:
        failed += 1

average = total / len(scores)

# ---------- گزارش ----------
print("\n===== گزارش کلاس =====")
print("نمرات وارد شده:", scores)
print("تعداد دانش‌آموزان:", len(scores))
print("مجموع نمرات:", total)
print("میانگین کلاس:", round(average, 2))
print("بالاترین نمره:", max(scores))
print("پایین‌ترین نمره:", min(scores))
print("تعداد قبولی:", passed)
print("تعداد مردودی:", failed)

# ---------- وضعیت هر نمره ----------
print("\n===== وضعیت تک‌تک =====")
for i in range(len(scores)):
    s = scores[i]
    if s >= 17:
        status = "عالی"
    elif s >= 14:
        status = "خوب"
    elif s >= 10:
        status = "قبول"
    else:
        status = "مردود"
    print("دانش‌آموز", i + 1, "→", s, "→", status)

# ---------- مرتب‌سازی ----------
scores.sort()
scores.reverse()
print("\nنمرات از زیاد به کم:", scores)
```

```python
numbers = [3, 8, 15, 22, 27, 30, 41, 50, 64, 71]
even_count=0
odd_count=0
for number in numbers:
    if number %2==0:
        print(number,"زوج کوچک")
        even_count=even_count+1
    elif number %2==0:
        print(number,"زوج بزرگ")
        even_count=even_count+1
    elif number %2==1:
        print(number,"فرد کوچک")
        odd_count=odd_count+1
    elif number %2==1:
        print(number,"فرد بزرگ")
```

```python
scores=[4, 9, 10, 12, 15, 16, 18, 20]
for score in scores:
    if 17<=score<=20 and score %2==0:
        print(score,"عالی","زوج")
    elif 17<=score<=20 and score %2!=0:
        print(score,"عالی","فرد")
    elif 14<=score<=16 and score %2==0:
        print(score,"خوب","زوج")
    elif 14<=score<=16 and score %2!=0:
        print(score,"خوب","فرد")
    elif 10<=score<=13 and score %2==0:
        print(score,"قابل قبول","زوج")
    elif 10<=score<=13 and score %2!=0:
        print(score,"قابل قبول","فرد")
    elif score <10 and score %2==0:
        print(score,"نیاز به تلاش بیشتر","زوج")
    elif score <10 and score %2!=0:
        print(score,"نیاز به تلاش بیشتر","فرد")
```
