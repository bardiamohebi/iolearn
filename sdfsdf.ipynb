```python
# ---------- گرفتن نمرات ----------
scores = []

n = int(input("چند تا دانش‌آموز داری؟ "))

for i in range(n):
    score = int(input("نمره‌ی دانش‌آموز شماره " + str(i + 1) + ": "))
    scores.append(score)

# ---------- محاسبات ----------
total = 0
passed = 0
failed = 0

for score in scores:
    total += score
    if score >= 10:
        passed += 1
    else:
        failed += 1

average = total / len(scores)

# ---------- گزارش ----------
print("\n===== گزارش کلاس =====")
print("نمرات وارد شده:", scores)
print("تعداد دانش‌آموزان:", len(scores))
print("مجموع نمرات:", total)
print("میانگین کلاس:", round(average, 2))
print("بالاترین نمره:", max(scores))
print("پایین‌ترین نمره:", min(scores))
print("تعداد قبولی:", passed)
print("تعداد مردودی:", failed)

# ---------- وضعیت هر نمره ----------
print("\n===== وضعیت تک‌تک =====")
for i in range(len(scores)):
    s = scores[i]
    if s >= 17:
        status = "عالی"
    elif s >= 14:
        status = "خوب"
    elif s >= 10:
        status = "قبول"
    else:
        status = "مردود"
    print("دانش‌آموز", i + 1, "→", s, "→", status)

# ---------- مرتب‌سازی ----------
scores.sort()
scores.reverse()
print("\nنمرات از زیاد به کم:", scores)
```
