```python
info={
    "name":"sara",
    "last name":"abasi",
    "age":14
}
print(info["name"])
info["age"]=24
info["color"]="black"
info.pop("name")
del info["color"]
print(info)
```

```python
if "last name" in info:
    print("این کلید وجود دارد")
else:
    print("این کلید وجود ندارد")
```

```python
if "color" not in info:
    print("متاسفانه موجود نیست")
else:
    print("موجود است")
```
