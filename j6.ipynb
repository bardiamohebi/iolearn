```python
def even_odd():
    number=int(input("عدد را وارد کنید"))
    if number  %2==0:
        print("عدد زوج")
    else:
        print("عدد فرد")
```

```python
def score():
    number=int(input("نمره را وارد کنید"))
    if 20>=number>=18:
        print(number,"خیلی خوب")
    elif 18>=number>=15:
        print(number,"خوب")
    elif 15>=number>=0:
        print(number,"تلاش کن")
    else:
        print(number,"نمره ی وارد شده معتبر نیست")
```

```python
score()
```
