```python
stocks=[0, 3, 7, 1, 15, 0, 5, 22, 2, 10]
d=0
e=0
f=0
g=0
for stock in stocks:
    if stock ==0:
        print(stock,"تمام شده")
        d=d+1
    elif 1<=stock<=5:
        print(stock,"کم")
        e=e+1
    elif 6<=stock<=10:
            print(stock,"مناسب")
            f=f+1
    elif stock >=10:
            print(stock,"موجودی زیاد")
            g=g+1
print("-----------")
print("موجودی زیاد",g)
print("مناسب",f)
print("کم",e)
print("تمام شده",d)
```
