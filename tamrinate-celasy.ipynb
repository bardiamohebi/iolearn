```python
kk=[4, 18, 7, 25, 30, 11, 42, 9, 16, 3]
even_count = 0
odd_count = 0
for k in kk:
    if k %2==0:
        even_count=even_count+1
        if k >=20:
         print(k,"عدد زوج بزرگ")
        elif k <=20:
           print(k,"عدد زوج کوچک")
    else:
       odd_count=odd_count+1
       if k >=20:
            print(k,"عدد فرد بزرگ")
       elif k <20:
            print(k,"عدد فرد کوچک")
print(even_count,"تعداد عداد زوج")
print(odd_count,"تعداد عداد فرد")
       
```
