```python
music=input("امروز چه حسی داری")
if music =="happy":
    print("موسیقسی pop")
elif music =="sad":
    print("موسیقی piano")
elif music =="enrgetic":
    print("موسیقی rock")
elif music =="calm":
    print("موسیقی lo-fi")
else:
    print("سبک مناسبی پدا نشد")
```

```python
Lesson=input("به کدام درس علاقه داری")
if Lesson =="ریاضی":
    print("شاید در اینده محندس شوی")
elif Lesson =="کامپیوتر":
    print("شاید برنامه نویس شوی")
elif Lesson =="زیست":
    print("شاید پزشک شوی")
elif Lesson =="هنر":
    print("شاید طراح گرافیک شوی")
else:
    print("هنوز فرصت داری علاقه ی واقی ات را پیدا کنی")
```

```python
cinma=int(input("سن خود را وارد کنید"))
if cinma <=7:
    print("ورود رایگان است")
elif 8<=cinma<=15:
    print("قیمت بلیت 80 هزار تومان است")
elif 16<=cinma<=60:
    print("قیمت بلیت 150 هزار تومان است")
elif 60<=cinma:
    print("قیمت بلیت 70 هزار تومان است")       
else:
    print("نمیدونم")
```

```python
battery=int(input("درصد باتری خود را وارد کنید"))
if battery <=20:
    print("شارژر خود را وصل کنید")
elif 20<=battery<=50:
    print("باتری در وضعیت متوسط قرار دارد")
elif battery >50:
    print("باتری در وضعیت عالی قرار دارد")
else:
    print("شارژ وارد شده معتبر نیست")
```

```python
onlineShop=int(input("مبلغ خرید خود را وارد کنید "))
if onlineShop <=300:
    print("هزینه ی ارسال 50 هزار تومان است")
elif 300<=onlineShop<=400:
    print("هزینه ی ارسال 20 هزار تومان است")
elif 400<=onlineShop<=700:
    print("ارسال رایگان است")
else:
    print("مبلغ وارد شده بیش از هد است")
```
